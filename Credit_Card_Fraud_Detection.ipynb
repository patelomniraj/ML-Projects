```python
import joblib
import numpy as np
from IPython.display import display, clear_output
import ipywidgets as widgets
import warnings
warnings.filterwarnings('ignore')

# Load model
model = joblib.load('fraud_detection_model.pkl')
expected_features = getattr(model, 'n_features_in_', 29)

# Load scaler if available
try:
    scaler = joblib.load('scaler.pkl')
    SCALER_AVAILABLE = True
except FileNotFoundError:
    scaler = None
    SCALER_AVAILABLE = False

# Define features based on model requirements
if expected_features == 29:
    FEATURES = ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10',
                'V11', 'V12', 'V13', 'V14', 'V15', 'V16', 'V17', 'V18', 'V19', 'V20',
                'V21', 'V22', 'V23', 'V24', 'V25', 'V26', 'V27', 'V28', 'Amount']
elif expected_features == 28:
    FEATURES = ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10',
                'V11', 'V12', 'V13', 'V14', 'V15', 'V16', 'V17', 'V18', 'V19', 'V20',
                'V21', 'V22', 'V23', 'V24', 'V25', 'V26', 'V27', 'V28']
else:
    FEATURES = [f'Feature_{i+1}' for i in range(expected_features)]

# Create widgets
feature_widgets = {}
for feature in FEATURES:
    value = 100.0 if feature == 'Amount' else 0.0
    feature_widgets[feature] = widgets.FloatText(
        value=value,
        description=f'{feature}:',
        style={'description_width': '60px'},
        layout=widgets.Layout(width='160px')
    )

predict_button = widgets.Button(description='Predict', button_style='success')
example_button = widgets.Button(description='Load Example', button_style='info')
clear_button = widgets.Button(description='Clear', button_style='warning')
output = widgets.Output()

def predict_fraud(button):
    with output:
        clear_output(wait=True)
        try:
            input_data = [feature_widgets[f].value for f in FEATURES]
            input_array = np.array(input_data).reshape(1, -1)

            if SCALER_AVAILABLE and scaler is not None:
                input_array = scaler.transform(input_array)

            prediction = model.predict(input_array)[0]

            try:
                probability = model.predict_proba(input_array)[0]
                prob_text = f" (Probability: {probability[1]:.1%})" if prediction == 1 else f" (Probability: {probability[0]:.1%})"
            except:
                prob_text = ""

            if prediction == 1:
                print(f"FRAUD DETECTED{prob_text}")
            else:
                print(f"LEGITIMATE TRANSACTION{prob_text}")

        except Exception as e:
            print(f"Error: {str(e)}")

def load_example(button):
    example_data = {
        'V1': -1.36, 'V2': -0.07, 'V3': 2.54, 'V4': 1.38, 'V5': -0.34,
        'V6': 0.46, 'V7': 0.24, 'V8': 0.10, 'V9': 0.36, 'V10': 0.09,
        'V11': -0.55, 'V12': -0.62, 'V13': -0.99, 'V14': -0.31, 'V15': 1.47,
        'V16': -0.47, 'V17': 0.21, 'V18': 0.03, 'V19': 0.40, 'V20': 0.25,
        'V21': -0.02, 'V22': 0.28, 'V23': -0.11, 'V24': 0.07, 'V25': 0.13,
        'V26': -0.19, 'V27': 0.13, 'V28': -0.02, 'Amount': 149.62
    }

    for feature in FEATURES:
        if feature in example_data:
            feature_widgets[feature].value = example_data[feature]

    with output:
        clear_output(wait=True)
        print("Example data loaded")

def clear_all(button):
    for feature in FEATURES:
        feature_widgets[feature].value = 100.0 if feature == 'Amount' else 0.0
    with output:
        clear_output(wait=True)
        print("Fields cleared")

predict_button.on_click(predict_fraud)
example_button.on_click(load_example)
clear_button.on_click(clear_all)

# Display interface
print("Credit Card Fraud Detection")
print(f"Features: {len(FEATURES)}")

# Display widgets in rows of 5
for i in range(0, len(FEATURES), 5):
    group = FEATURES[i:i+5]
    widgets_row = [feature_widgets[f] for f in group]
    display(widgets.HBox(widgets_row))

# Controls
display(widgets.HBox([predict_button, example_button, clear_button]))
display(output)
```
