Cleaning

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download("stopwords")
```

```python
list1 = stopwords.words('english')
l = [
    '<', '>', '/', '.', ',', ':', '!', '?', '(', ')',
    '[', ']', '{', '}', '...', 'br', '..', "'", '"',
    '|', '\\', ';', '-', '&', '#', '!', '@', '%', '``',
    "'s", ' ', '*', '', '_', '^', '~', '+', '=', '$','.....','....','--'
]
list1.extend(l)

list1
```

```python
import re
```

```python
from nltk.stem import PorterStemmer
porter = PorterStemmer()
```

```python
def cleaning_stem(arr):

  for i in arr:
    # Create a token
    token_array = sent_tokenize(i)
    #print(token_array)

    # word token
    for i in range(len(token_array)):
      words  = word_tokenize(token_array[i])
      #print(words)

      words = [ porter.stem(word) for word in words if word not in set(list1)]

      token_array[i] = ' '.join(words) # Converting list of words into Sentences and placing back into sentence list

    print(token_array)
```

```python
cleaning_stem(corpus)
```

# Lemmatization

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemma = WordNetLemmatizer()
```

```python
import re
def cleaning_lemma(arr):

  for i in arr:
    # Create a token
    token_array = sent_tokenize(i)
    #print(token_array)

    # word token
    for i in range(len(token_array)):
      words  = word_tokenize(token_array[i].lower())
      #print(words)

      words = [ lemma.lemmatize(word,pos='v') for word in words if word not in set(list1)]

      token_array[i] = ' '.join(words) # Converting list of words into Sentences and placing back into sentence list

    print(token_array)
```

```python
cleaning_lemma(corpus)
```
